# Synthetic Insurance Claims Image Dataset Generation

Generate 24 photorealistic insurance claim photographs (`images/claim_001.jpg` through `images/claim_024.jpg`) and a ground-truth manifest (`manifest.json`) for the [Multimodal Insurance Claims Adjustment workflow](../../workflows/claims_adjustment/).

These images are designed to test a 5-stage multimodal BigQuery AI pipeline (`AI.SCORE`, `AI.IF`, `AI.CLASSIFY`, `AI.GENERATE`, `AI.AGG`):
1. **Auto Claims — Clear & Sub-Classifiable (`8` photos)** — `collision_panel` (`3`), `glass_windshield` (`3`), and `hail_dents` (`2`)
2. **Property / Roof Claims — Clear & Sub-Classifiable (`8` photos)** — `wind_missing_shingles` (`3`), `hail_impact` (`2`), and `tree_structural_impact` (`3`)
3. **Clear Photos Requiring Human Specialist Triage (`4` photos)**:
   - *Unclassified Type at Step 2 (`2` photos)*: Clear photos of non-auto, non-roof property damage (flooded basement water heater; storm-felled wooden backyard fence) $\rightarrow$ routed to `'unclassified'` at Step 2
   - *Unclear Subtype at Step 3 (`2` photos)*: Clear photos of an atypical auto claim (shredded tire blowout on wheel rim) and an atypical roof claim (rusted chimney flashing & satellite dish bracket with intact shingles) $\rightarrow$ routed to `'other_unclear'` at Step 3
4. **Low-Quality / Uninspectable Photos (`4` photos)** — severe motion blur, pitch-dark night exposure, extreme lens defocus, and blurry non-asset paperwork obstructed by a finger $\rightarrow$ stopped at Step 1 (`Claimant Re-Upload Queue`) before spending tokens on classification or extraction

**Generic Shuffled Naming (`claim_001.jpg`–`claim_024.jpg`):** All 24 specifications are deterministically shuffled (`SEED = 42`) before assigning `claim_NNN.jpg` filenames so filenames never leak category, subtype, or image quality to BigQuery AI functions.

**Prerequisites:** A GCP project with Vertex AI enabled. See the [Setup Reference](../../setup/) for details.

## Step 2 — Generate Photorealistic Claim Photos via Vertex AI & Apply Controlled Quality Defects

We call Vertex AI's Gemini image generation model (`gemini-2.5-flash-image`) to synthesize each base photograph, resize to `640×480` RGB, and apply deterministic `Pillow` post-processing to the 4 low-quality intake test cases so their visual defects (`motion_blur`, `pitch_dark`, `extreme_defocus`, `blurry_obstruction`) are reproducible across runs.

Before writing files to disk, we shuffle the 24 specifications with `random.Random(SEED)` and assign generic filenames (`claim_001.jpg` through `claim_024.jpg`).

```python
import base64
import io
import json
import random
import time
import numpy as np
from PIL import Image, ImageDraw, ImageEnhance, ImageFilter
import google.auth
from google.auth.transport.requests import AuthorizedSession

creds, _ = google.auth.default(scopes=['https://www.googleapis.com/auth/cloud-platform'])
authed_session = AuthorizedSession(creds)
IMAGE_GEN_NAME = 'gemini-2.5-flash-image'
GEN_URL = (
    f'https://aiplatform.googleapis.com/v1/projects/{PROJECT_ID}'
    f'/locations/{LOCATION}/publishers/google/models/{IMAGE_GEN_NAME}:generateContent'
)


def generate_raw_image(prompt: str, max_retries: int = 4) -> Image.Image:
    """Generate a photorealistic image via Vertex AI Gemini image generation."""
    payload = {
        'contents': [{'role': 'user', 'parts': [{'text': prompt}]}],
        'generationConfig': {'responseModalities': ['TEXT', 'IMAGE']},
    }
    for attempt in range(max_retries):
        resp = authed_session.post(GEN_URL, json=payload, timeout=90)
        if resp.status_code == 200:
            candidates = resp.json().get('candidates', [])
            for cand in candidates:
                for part in cand.get('content', {}).get('parts', []):
                    if 'inlineData' in part:
                        raw_bytes = base64.b64decode(part['inlineData']['data'])
                        return Image.open(io.BytesIO(raw_bytes)).convert('RGB')
        time.sleep(2.0 * (attempt + 1))
    raise RuntimeError(f'Image generation failed after {max_retries} attempts: {resp.status_code} {resp.text[:200]}')


def apply_postprocess(img: Image.Image, mode: str) -> Image.Image:
    """Resize to 640x480 (4:3) and apply deterministic quality degradation when requested."""
    # Center-crop to 4:3 aspect ratio, then resize to 640x480
    w, h = img.size
    target_ratio = 4.0 / 3.0
    if (w / h) > target_ratio:
        new_w = int(h * target_ratio)
        left = (w - new_w) // 2
        img = img.crop((left, 0, left + new_w, h))
    else:
        new_h = int(w / target_ratio)
        top = (h - new_h) // 2
        img = img.crop((0, top, w, top + new_h))
    img = img.resize((640, 480), Image.Resampling.LANCZOS)

    if mode == 'none':
        return img
    if mode == 'motion_blur':
        # Heavy directional horizontal motion blur + Gaussian blur
        arr = np.array(img, dtype=np.float32)
        shifted = np.zeros_like(arr)
        shifts = 45
        for s in range(shifts):
            shifted += np.roll(arr, s * 4, axis=1)
        shifted /= shifts
        blurred = Image.fromarray(np.clip(shifted, 0, 255).astype(np.uint8))
        return blurred.filter(ImageFilter.GaussianBlur(radius=10))
    if mode == 'pitch_dark':
        # Extreme underexposure (4% brightness) + low contrast
        dark = ImageEnhance.Brightness(img).enhance(0.04)
        return dark.filter(ImageFilter.GaussianBlur(radius=6))
    if mode == 'extreme_defocus':
        # Heavy lens defocus (radius=26 Gaussian blur)
        return img.filter(ImageFilter.GaussianBlur(radius=26))
    if mode == 'blurry_obstruction':
        # Out-of-focus non-asset scene with a large blurry thumb covering the lens
        blurred = img.filter(ImageFilter.GaussianBlur(radius=14))
        draw = ImageDraw.Draw(blurred)
        draw.ellipse([220, 80, 660, 500], fill=(198, 142, 122))
        return blurred.filter(ImageFilter.GaussianBlur(radius=18))
    return img


# Deterministically shuffle specifications so claim_001.jpg..claim_024.jpg never leak category
shuffled_specs = list(CLAIM_SPECS)
rng = random.Random(SEED)
rng.shuffle(shuffled_specs)

manifest_records = []
print(f'Generating {len(shuffled_specs)} claim images in {IMAGES_DIR}...')
for idx, spec in enumerate(shuffled_specs, start=1):
    claim_id = f'claim_{idx:03d}'
    filename = f'{claim_id}.jpg'
    filepath = IMAGES_DIR / filename

    raw_img = generate_raw_image(spec['prompt'])
    final_img = apply_postprocess(raw_img, spec['postprocess'])

    buf = io.BytesIO()
    final_img.save(buf, format='JPEG', quality=82, optimize=True)
    filepath.write_bytes(buf.getvalue())

    record = {
        'claim_id': claim_id,
        'filename': filename,
        'spec_id': spec['spec_id'],
        'quality_pass': spec['quality_pass'],
        'quality_defect': spec['quality_defect'],
        'claim_type': spec['claim_type'],
        'claim_subtype': spec['claim_subtype'],
        'expected_routing': spec['expected_routing'],
        'ground_truth_severity_tier': spec['ground_truth_severity_tier'],
        'description': spec['description'],
        'size_kb': round(len(buf.getvalue()) / 1024.0, 1),
    }
    manifest_records.append(record)
    print(f'  [{idx:02d}/24] {filename} <- {spec["spec_id"]:24s} ({record["size_kb"]:4.1f} KB)')

MANIFEST_PATH.write_text(json.dumps(manifest_records, indent=2) + '\n')
total_kb = sum(r['size_kb'] for r in manifest_records)
print(f'\nSaved 24 JPEG images ({total_kb:.1f} KB total) and {MANIFEST_PATH.name}')
```
