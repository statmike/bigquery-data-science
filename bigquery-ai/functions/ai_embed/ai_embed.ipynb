# AI.EMBED — BigQuery AI Functions

`AI.EMBED` is a scalar function that creates text or multimodal embeddings using `gemini-embedding-2` (or other Vertex AI endpoints), or using the built-in `embeddinggemma-300m` model with no Vertex AI charges — no model creation required either way.

**When to use it:**
- You need embeddings for individual values in SELECT, WHERE, or JOIN
- You want to create text or multimodal embeddings without CREATE MODEL
- You need to embed a small number of values inline
- You want zero-cost text embedding using the built-in model (`embeddinggemma-300m`)

**Alternatives:**
- [`AI.GENERATE_EMBEDDING`](../ai_generate_embedding/) — TVF for batch embedding, requires CREATE MODEL
- [`AI.SIMILARITY`](../ai_similarity/) — Cosine similarity between two inputs (uses embeddings internally)
- [`VECTOR_SEARCH`](../vector_search/) — Top-K nearest neighbor search on pre-computed embeddings

**Featured in:** [Semantic Search System](../../workflows/semantic_search/) | [RAG Pipeline](../../workflows/rag_pipeline/) | [Multimodal Analysis](../../workflows/multimodal_analysis/) | [Document RAG Pipeline](../../workflows/document_rag/) | [Log Analysis](../../workflows/log_analysis/) | [Catalog Search](../../workflows/catalog_search/) | [Image Deduplication](../../workflows/image_deduplication/)

**Multimodal:** `gemini-embedding-2` (GA) is the primary unified model for text, images, PDFs, audio, and video via [ObjectRef](../../reference/unstructured-data-infrastructure.md#objectref-and-objectrefruntime-schema-reference) (3,072 dimensions by default, reducible via `model_params => JSON '{"outputDimensionality": 768}'`). Legacy `multimodalembedding@001` (1,408 dimensions) also supports images and video.

**References:** [Full syntax reference](../../RESOURCES.md) | [Official documentation](https://cloud.google.com/bigquery/docs/reference/standard-sql/bigqueryml-syntax-ai-embed) | [Setup guide](../../setup/)

## Examples — SQL

Progressive examples from simplest to most advanced. Each cell adds one new concept.

### 1. Basic text embedding

`AI.EMBED` takes content and an endpoint, returns a STRUCT with `result` (the embedding vector) and `status`.

**Note:** Either `endpoint` or `model` is required. When using Vertex AI models, `endpoint` must include the model version. When using the built-in model, use `model` instead (see Example 6).

```python
query = """
SELECT
  text,
  ARRAY_LENGTH((AI.EMBED(
    content => text,
    endpoint => 'gemini-embedding-2'
  )).result) AS embedding_dimensions
FROM UNNEST(['BigQuery is a data warehouse.', 'Cloud computing scales on demand.']) AS text
"""
client.query(query).to_dataframe()
```

### 2. Viewing the embedding vector

The embedding is an `ARRAY<FLOAT64>` — a dense vector of floating-point numbers.

```python
query = """
SELECT
  (AI.EMBED(
    content => 'What is machine learning?',
    endpoint => 'gemini-embedding-2'
  )).result AS embedding
"""
df = client.query(query).to_dataframe()
embedding = df.iloc[0]['embedding']
print(f'Dimensions: {len(embedding)}')
print(f'First 5 values: {embedding[:5]}')
```

### 3. Using `task_type` and `title` (legacy text models)

> **Important (`gemini-embedding-2` vs legacy models):** `gemini-embedding-2` produces unified, task-agnostic embeddings across text and multimodal inputs and **rejects `task_type` and `title`** (passing either raises `400 The task_type and title arguments of ai.embed are not supported for gemini-embedding-2 model`). Only legacy text embedding models (`gemini-embedding-001` and `text-embedding-005`) accept `task_type` and `title`.

On `gemini-embedding-001` and `text-embedding-005`, `task_type` tells the embedding model the intended use of the text:

**Asymmetric task types** — documents and queries use *different* task types:

| Use Case | Document Task Type | Query Task Type |
|----------|-------------------|-----------------|
| Search | `RETRIEVAL_DOCUMENT` | `RETRIEVAL_QUERY` |
| Question Answering | `RETRIEVAL_DOCUMENT` | `QUESTION_ANSWERING` |
| Fact Checking | `RETRIEVAL_DOCUMENT` | `FACT_VERIFICATION` |
| Code Retrieval | `RETRIEVAL_DOCUMENT` | `CODE_RETRIEVAL_QUERY` |

**Symmetric task types** — both sides use the *same* task type:
- `SEMANTIC_SIMILARITY` — comparing how similar two texts are
- `CLASSIFICATION` — grouping texts by category
- `CLUSTERING` — organizing texts into clusters

When `task_type => 'RETRIEVAL_DOCUMENT'` is used on `gemini-embedding-001` or `text-embedding-005`, you can also pass an optional `title`:

```python
query = """
SELECT
  text,
  ARRAY_LENGTH((AI.EMBED(
    content => text,
    endpoint => 'gemini-embedding-001',
    task_type => 'RETRIEVAL_DOCUMENT',
    title => 'Google Cloud Service Overview'
  )).result) AS gemini_001_dims,
  ARRAY_LENGTH((AI.EMBED(
    content => text,
    endpoint => 'text-embedding-005',
    task_type => 'RETRIEVAL_DOCUMENT',
    title => 'Google Cloud Service Overview'
  )).result) AS text_005_dims
FROM UNNEST([
  'BigQuery is a serverless data warehouse.',
  'Cloud Functions runs event-driven code.',
  'Cloud Storage stores objects in buckets.'
]) AS text
"""
client.query(query).to_dataframe()
```

### 4. Dimension reduction with `model_params` and cosine similarity

`gemini-embedding-2` defaults to `3,072` dimensions, whereas BigQuery `CREATE VECTOR INDEX` supports at most `2,000` dimensions. Pass `model_params => JSON '{"outputDimensionality": 768}'` on `AI.EMBED` to reduce `gemini-embedding-2` vectors to `768` Matryoshka dimensions so they can be indexed and compared efficiently:

> **Pitfall:** The JSON key in `AI.EMBED` must be camelCase `"outputDimensionality"`. Passing `"dimension"` or `"output_dimensionality"` inside `model_params` on `AI.EMBED` is ignored by `gemini-embedding-2` and still returns `3,072` dimensions.

```python
query = f'''
WITH embeddings AS (
  SELECT
    text,
    (AI.EMBED(
      content => text,
      endpoint => 'gemini-embedding-2',
      model_params => JSON '{{"outputDimensionality": 768}}'
    )).result AS vec
  FROM UNNEST([
    'BigQuery is a data warehouse',
    'BigQuery stores and analyzes large datasets',
    'I enjoy hiking in the mountains'
  ]) AS text
)
SELECT
  a.text AS text_a,
  b.text AS text_b,
  ARRAY_LENGTH(a.vec) AS dims,
  1 - ML.DISTANCE(a.vec, b.vec, 'COSINE') AS cosine_similarity
FROM embeddings a
CROSS JOIN embeddings b
WHERE a.text < b.text
ORDER BY cosine_similarity DESC
'''
client.query(query).to_dataframe()
```

### 5. Saving embeddings to a table

For reuse across queries, materialize embeddings into a table.

```python
query = f'''
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.ai_embed_documents` AS
SELECT
  id,
  text,
  (AI.EMBED(
    content => text,
    endpoint => 'gemini-embedding-2',
    model_params => JSON '{{"outputDimensionality": 768}}'
  )).result AS embedding
FROM UNNEST([
  STRUCT(1 AS id, 'BigQuery is a serverless data warehouse for analytics.' AS text),
  STRUCT(2, 'Cloud Functions lets you run code in response to events.'),
  STRUCT(3, 'Cloud Storage provides object storage for any amount of data.'),
  STRUCT(4, 'Kubernetes Engine runs containerized applications at scale.'),
  STRUCT(5, 'Pub/Sub is a messaging service for event-driven systems.')
])
'''
client.query(query).result()

# Verify
client.query(f'SELECT id, text, ARRAY_LENGTH(embedding) AS dims FROM `{PROJECT_ID}.{DATASET_ID}.ai_embed_documents` ORDER BY id').to_dataframe()
```

### 6. Built-in model (no Vertex AI charges)

Use `model => 'embeddinggemma-300m'` for zero-cost text embedding. Data stays in BigQuery and uses BQ compute slots — no Vertex AI endpoint call, no connection needed, no per-token billing. Returns 768-dimension vectors.

This is ideal for cost-sensitive workloads or when you want to avoid Vertex AI dependencies entirely.

> **Note:** `embeddinggemma-300m` is in Preview. When using `model`, you cannot specify `endpoint`, `title`, `model_params`, or `connection_id`.

```python
query = """
SELECT
  text,
  ARRAY_LENGTH((AI.EMBED(
    content => text,
    model => 'embeddinggemma-300m'
  )).result) AS embedding_dimensions
FROM UNNEST(['BigQuery is a data warehouse.', 'Cloud computing scales on demand.']) AS text
"""
client.query(query).to_dataframe()
```

### 7. Comparing embedding models

`AI.EMBED` supports multiple embedding models with different dimensions, capabilities, and parameter support:

| Model | Param | Type | Default Dims | Notes |
|-------|-------|------|--------------|-------|
| `gemini-embedding-2` | `endpoint` | Vertex AI unified text + multimodal (images, PDFs, audio, video) | 3072 | Primary GA model; task-agnostic (rejects `task_type`/`title`); reducible via `{"outputDimensionality": 768}` |
| `gemini-embedding-001` | `endpoint` | Vertex AI text | 3072 | Legacy multilingual text model; accepts `task_type` and `title` |
| `text-embedding-005` | `endpoint` | Vertex AI text | 768 | Legacy text model; accepts `task_type` and `title` |
| `embeddinggemma-300m` | `model` | Built-in text | 768 | Zero Vertex AI charges (BQ slots only) |

```python
query = """
SELECT 'gemini-embedding-2 (primary)' AS model,
  ARRAY_LENGTH((AI.EMBED(content => 'BigQuery is a data warehouse', endpoint => 'gemini-embedding-2')).result) AS dims
UNION ALL
SELECT 'gemini-embedding-001',
  ARRAY_LENGTH((AI.EMBED(content => 'BigQuery is a data warehouse', endpoint => 'gemini-embedding-001')).result)
UNION ALL
SELECT 'text-embedding-005',
  ARRAY_LENGTH((AI.EMBED(content => 'BigQuery is a data warehouse', endpoint => 'text-embedding-005')).result)
UNION ALL
SELECT 'embeddinggemma-300m (built-in)',
  ARRAY_LENGTH((AI.EMBED(content => 'BigQuery is a data warehouse', model => 'embeddinggemma-300m')).result)
"""
client.query(query).to_dataframe()
```

## Multimodal — Document Embedding

`AI.EMBED` supports multimodal inputs (images, PDFs, audio, video, and mixed STRUCTs) via ObjectRef using `gemini-embedding-2` (GA). A `connection_id` is required for multimodal inputs. `gemini-embedding-2` returns 3,072-dimension vectors by default (reducible via `model_params => JSON '{"outputDimensionality": 768}'`).

> **Note:** Legacy `multimodalembedding@001` (1,408 dimensions) also supports images and video, while `gemini-embedding-2` serves as the unified text and multimodal model across text, images, PDFs, audio, and video. See [Embedding Generation and Semantic Search](../../reference/embedding-generation-and-semantic-search.md) for details.

Below we render invoice and receipt PDFs from this project's [document set](../../data/documents/) to PNG images, upload them to GCS, then embed and compare them with `gemini-embedding-2`. The embeddings show that documents of the same type cluster together.

The ObjectRef pattern: `OBJ.MAKE_REF(uri, connection)` builds the reference inline — no object table needed, and nothing to sign.

```python
import shutil, subprocess

if not shutil.which('pdftoppm'):
    subprocess.check_call(['sudo', 'apt-get', 'install', '-y', '-qq', 'poppler-utils'])
    print('Installed poppler-utils (provides pdftoppm)')
else:
    print('pdftoppm already available')
```

```python
import subprocess, io
from pathlib import Path
from google.cloud import storage

# Locate the document directory
data_dir = Path('../../data/documents')
if not data_dir.exists():
    data_dir = Path('data/documents')

# Pick 3 invoices and 3 receipts
docs = [
    ('invoice_1.png', data_dir / 'invoices' / 'invoice_001.pdf'),
    ('invoice_2.png', data_dir / 'invoices' / 'invoice_002.pdf'),
    ('invoice_3.png', data_dir / 'invoices' / 'invoice_003.pdf'),
    ('receipt_1.png', data_dir / 'receipts' / 'receipt_001.pdf'),
    ('receipt_2.png', data_dir / 'receipts' / 'receipt_002.pdf'),
    ('receipt_3.png', data_dir / 'receipts' / 'receipt_003.pdf'),
]

gcs = storage.Client(project=PROJECT_ID)
bucket = gcs.bucket(BUCKET)
prefix = 'bq_ai_functions/ai_embed'

for name, pdf_path in docs:
    # Render first page of PDF to PNG using pdftoppm (poppler)
    result = subprocess.run(
        ['pdftoppm', '-png', '-f', '1', '-l', '1', '-r', '150', str(pdf_path)],
        capture_output=True
    )
    blob = bucket.blob(f'{prefix}/{name}')
    blob.upload_from_string(result.stdout, content_type='image/png')

print(f'Rendered and uploaded {len(docs)} document images to gs://{BUCKET}/{prefix}/')
```

### 8. Embed a document image

Use an inline `OBJ.MAKE_REF` to embed an invoice image with `gemini-embedding-2`. The `connection_id` parameter authorizes GCS access.

```python
query = f"""
SELECT
  'invoice_1.png' AS document,
  ARRAY_LENGTH((AI.EMBED(
    content => OBJ.MAKE_REF(
      'gs://{BUCKET}/bq_ai_functions/ai_embed/invoice_1.png',
      '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}'
    ),
    endpoint => 'gemini-embedding-2',
    connection_id => '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}'
  )).result) AS embedding_dimensions
"""
client.query(query).to_dataframe()
```

### 9. Compare document embeddings

Embed all 6 documents and compute pairwise cosine similarity using `ML.DISTANCE`. Documents of the same type (invoices vs receipts) should cluster together with higher similarity scores.

```python
query = f"""
WITH doc_embeddings AS (
  SELECT
    doc_name,
    (AI.EMBED(
      content => OBJ.MAKE_REF(
        CONCAT('gs://{BUCKET}/bq_ai_functions/ai_embed/', doc_name),
        '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}'
      ),
      endpoint => 'gemini-embedding-2',
      connection_id => '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}'
    )).result AS vec
  FROM UNNEST([
    'invoice_1.png', 'invoice_2.png', 'invoice_3.png',
    'receipt_1.png', 'receipt_2.png', 'receipt_3.png'
  ]) AS doc_name
)
SELECT
  a.doc_name AS doc_a,
  b.doc_name AS doc_b,
  ROUND(1 - ML.DISTANCE(a.vec, b.vec, 'COSINE'), 4) AS cosine_similarity
FROM doc_embeddings a
CROSS JOIN doc_embeddings b
WHERE a.doc_name < b.doc_name
ORDER BY cosine_similarity DESC
"""
client.query(query).to_dataframe()
```
