# BigQuery AI Functions — Overview

This notebook gives you a runnable tour of BigQuery's AI functions — working examples across every category, plus GoogleSQL's reusable column aliases for composing multi-step `AI.*` pipelines cleanly.

**Sections covered:**
1. **Generation** — Generate text and structured output with `AI.GENERATE` (GA)
2. **Managed Functions** — Classify, score, filter, and translate with `AI.CLASSIFY`, `AI.SCORE`, `AI.IF`, and `AI.TRANSLATE` (plus aggregate `AI.AGG` — all GA)
3. **Structured Output** — Generate tabular data with `AI.GENERATE_TABLE` (GA)
4. **Embeddings & Search** — Create embeddings with `AI.EMBED` (GA) and search with `VECTOR_SEARCH` (GA; hybrid search Preview)
5. **Document Processing** — Chunk PDFs and Office/HTML files in Cloud Storage with zero DDL using `AI.CHUNK_DOCUMENT` (GA)
6. **Predictive AI** — Forecast time series with `AI.FORECAST` (GA) and predict tabular outcomes with `AI.PREDICT` (Preview)
7. **Augmented Analytics** — Explain what moved a metric with `AI.KEY_DRIVERS` (GA)
8. **Utility** — Size and cost text prompts for free before sending them with `AI.COUNT_TOKENS` (GA)
9. **SQL Ergonomics — Reusable Column Aliases** — Chain `AI.*` scalar calls and unpack `STRUCT` outputs left-to-right in a single `SELECT` (GA)

**For deeper examples** of any function (including pre-trained Cloud AI Service Models such as `ML.PROCESS_DOCUMENT` and `ML.TRANSLATE`), see the individual notebooks in [`functions/`](functions/) and end-to-end compositions in [`workflows/`](workflows/).

**For setup details** (connections, models, quotas, IAM), see the [Setup Reference](setup/).

## 1. Generation — `AI.GENERATE`

`AI.GENERATE` (GA) is the simplest way to call a Gemini model from SQL. It is a **scalar function** — use it in `SELECT`, `WHERE`, or anywhere you would use a regular SQL function, with no `CREATE MODEL` step required.

**Deep dive:** [`functions/ai_generate/`](functions/ai_generate/)

````python
%%bigquery --project {PROJECT_ID}

-- Simplest possible call: just a prompt string
SELECT
  city,
  (AI.GENERATE(CONCAT('What country is ', city, ' in? Answer in one word.'))).result AS country
FROM UNNEST(['Tokyo', 'Paris', 'Nairobi', 'Lima']) AS city
````

### Structured output with `output_schema`

`AI.GENERATE` can return typed columns instead of free-form text when you provide an `output_schema`.

````python
%%bigquery --project {PROJECT_ID}

-- Structured output: get typed columns back
SELECT
  animal,
  result.habitat,
  result.average_lifespan_years,
  result.is_endangered
FROM
  UNNEST(['Eagle', 'Salmon', 'Cobra']) AS animal,
  UNNEST([
    AI.GENERATE(
      CONCAT('Give me facts about: ', animal),
      output_schema => 'habitat STRING, average_lifespan_years INT64, is_endangered BOOL'
    )
  ]) AS result
````

## 2. Managed Functions — `AI.CLASSIFY`, `AI.SCORE`, `AI.IF`, `AI.TRANSLATE`

Managed functions (all **GA**, along with the aggregate [`AI.AGG`](functions/ai_agg/)) provide **task-specific interfaces** with zero `CREATE MODEL` DDL:
- `AI.CLASSIFY`, `AI.SCORE`, and `AI.IF` call Gemini with automatic prompt structuring and return plain scalar values (`STRING` / `ARRAY<STRING>`, `FLOAT64`, `BOOL`) rather than structs. *(Embedding-distilled `optimization_mode` on `AI.IF` and `AI.CLASSIFY` is in Preview.)*
- `AI.TRANSLATE` calls Cloud Translation Advanced (`model => 'nmt'` default or `'translation-llm'`) under caller end-user credentials.

**Deep dives:** [`functions/ai_classify/`](functions/ai_classify/) | [`functions/ai_score/`](functions/ai_score/) | [`functions/ai_if/`](functions/ai_if/) | [`functions/ai_translate/`](functions/ai_translate/) | [`functions/ai_agg/`](functions/ai_agg/)

````python
%%bigquery --project {PROJECT_ID}

-- AI.CLASSIFY: assign a category from a list you provide
-- AI.SCORE: rate on a scale you describe
-- AI.IF: evaluate a natural language condition to TRUE/FALSE
SELECT
  review,
  AI.CLASSIFY(
    review,
    ['Product Quality', 'Shipping', 'Customer Service', 'Pricing']
  ) AS category,
  AI.SCORE(
    CONCAT('Rate the sentiment of this review on a scale from 1 (very negative) to 5 (very positive): ', review)
  ) AS sentiment_score,
  AI.IF(
    CONCAT(review, ' — Does this review mention a refund request?')
  ) AS mentions_refund
FROM UNNEST([
  'The product broke after two days. I want my money back!',
  'Shipping was incredibly fast, arrived next day.',
  'Great price for the quality. Will buy again.',
  'Customer support was unhelpful and rude.'
]) AS review
````

### Multilingual translation with `AI.TRANSLATE`

`AI.TRANSLATE` (GA) translates text per row with zero `CREATE MODEL` or connection setup, and — unlike [`ML.TRANSLATE`](functions/ml_translate/) — accepts **dynamic per-row `target_language` and `source_language` column expressions**.

- **Engine choice:** `model => 'nmt'` (Cloud Translation Neural Machine Translation, default) or `model => 'translation-llm'` (Cloud Translation LLM for idiomatic nuance).
- **Row-level diagnostics (`return_details => TRUE`):** Changes the return type from `STRING` (`NULL` on error) to `STRUCT<result STRING, status STRING, details JSON>`. Successful rows emit **`status = 'OK'`** (not empty string `''`, which `AI.GENERATE` and `ML.TRANSLATE` use), and `details` exposes snake_case `detected_language_code`.
- **Auth & residency:** Runs under caller end-user credentials (`roles/cloudtranslate.user`; `connection_id` is not accepted). Queries in `US` or `EU` multi-regions stay in `us` or `eu`, whereas single-region queries route to the `global` Cloud Translation endpoint — use [`ML.TRANSLATE`](functions/ml_translate/) when you require connection service-account auth or standalone language detection/transliteration.

````python
%%bigquery --project {PROJECT_ID}

-- Translate each row into its own target_language and inspect auto-detected source language
WITH multilingual_inputs AS (
  SELECT * FROM UNNEST([
    STRUCT(1 AS id, 'La livraison a pris deux semaines de retard.' AS source_text, 'en' AS target_lang),
    STRUCT(2 AS id, 'Your replacement sensor shipped this morning.', 'es'),
    STRUCT(3 AS id, 'Die Rechnung stimmt nicht mit dem Angebot überein.', 'ja')
  ])
),
translated AS (
  SELECT
    id,
    source_text,
    target_lang,
    AI.TRANSLATE(
      source_text,
      target_lang,
      model => 'nmt',
      return_details => TRUE
    ) AS tr
  FROM multilingual_inputs
)
SELECT
  id,
  target_lang,
  JSON_VALUE(tr.details, '$.detected_language_code') AS detected_source_lang,
  tr.result AS translated_text,
  tr.status
FROM translated
WHERE tr.status = 'OK'  -- AI.TRANSLATE uses 'OK' on success (not '')
ORDER BY id
````

## 3. Structured Output — `AI.GENERATE_TABLE`

`AI.GENERATE_TABLE` is a **table-valued function** that returns structured columns defined by an `output_schema`. It requires a connection and remote model — both are created below.

**One output row per input prompt row.** The function calls the model once for each row of the input relation and returns one row per call, so the row count comes from the input, never from the prompt text: a single prompt asking for ten records returns one row containing one record. To get ten rows, pass ten prompts — the example below drives them from `UNNEST(GENERATE_ARRAY(1, 10))`. Input columns pass through, so keep the generator column (`n` here) to order and join on.

This is particularly useful for **generating sample data** — which is how we use it throughout this project to keep examples self-contained.

**Deep dive:** [`functions/ai_generate_table/`](functions/ai_generate_table/)

````python
import re, subprocess, json

# Create connection (idempotent — needed for AI.GENERATE_TABLE)
subprocess.run(
    ['bq', 'mk', '--connection', '--location', LOCATION,
     '--connection_type', 'CLOUD_RESOURCE',
     '--project_id', PROJECT_ID, CONNECTION_ID],
    capture_output=True, text=True
)

# Grant Vertex AI User role to connection's service account (idempotent)
r = subprocess.run(
    ['bq', 'show', '--connection', '--format=json',
     '--project_id', PROJECT_ID, '--location', LOCATION, CONNECTION_ID],
    capture_output=True, text=True, check=True
)
sa = json.loads(r.stdout)['cloudResource']['serviceAccountId']
subprocess.run(
    ['gcloud', 'projects', 'add-iam-policy-binding', PROJECT_ID,
     f'--member=serviceAccount:{sa}', '--role=roles/aiplatform.user', '--quiet'],
    capture_output=True, text=True
)
sa_display = re.sub(r'bqcx-\d+-', 'bqcx-000000000000-', sa)
print(f'Connection {CONNECTION_ID} ready (service account: {sa_display})')

# Create remote model (idempotent)
create_model_sql = f"""
CREATE OR REPLACE MODEL `{PROJECT_ID}.{DATASET_ID}.gemini_flash`
  REMOTE WITH CONNECTION `{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}`
  OPTIONS (endpoint = 'gemini-2.5-flash')
"""
client.query(create_model_sql).result()
print('Model gemini_flash ready')
````

````python
N_TICKETS = 10  # row count comes from the input relation, not from the prompt text

query = f"""
SELECT n, ticket_id, customer_name, department, ticket_text
FROM AI.GENERATE_TABLE(
  MODEL `{PROJECT_ID}.{DATASET_ID}.gemini_flash`,
  (
    -- One prompt row per ticket: N input rows in, N output rows out
    SELECT
      n,
      CONCAT(
        'Write one fictional customer support ticket for an online retailer. ',
        'It is a ', ['complaint', 'question', 'compliment'][ORDINAL(MOD(n, 3) + 1)], '. ',
        'Use ticket_id ', CAST(1000 + n AS STRING), '. ',
        'Make the text realistic and vary the length.'
      ) AS prompt
    FROM UNNEST(GENERATE_ARRAY(1, {N_TICKETS})) AS n
  ),
  STRUCT(
    'ticket_id INT64, customer_name STRING, ticket_text STRING, department STRING' AS output_schema
  )
)
ORDER BY n
"""
df_tickets = client.query(query).to_dataframe()
print(f'Prompt rows in: {N_TICKETS}   |   Generated rows out: {len(df_tickets)}')
df_tickets
````

## 4. Embeddings & Search — `AI.EMBED` + `VECTOR_SEARCH`

`AI.EMBED` (GA, alongside pairwise [`AI.SIMILARITY`](functions/ai_similarity/)) creates vector embeddings from text or `ObjectRef` media as a scalar function with no `CREATE MODEL` step. `VECTOR_SEARCH` (GA; single-query and hybrid search syntax Preview) finds nearest neighbors from pre-computed embeddings stored in a table.

**Deep dives:** [`functions/ai_embed/`](functions/ai_embed/) | [`functions/vector_search/`](functions/vector_search/) | [`functions/ai_search/`](functions/ai_search/)

````python
%%bigquery --project {PROJECT_ID}

-- AI.EMBED returns a vector embedding for each input text
SELECT
  doc,
  ARRAY_LENGTH(embedding) AS dimensions
FROM (
  SELECT
    doc,
    (AI.EMBED(
      content => doc,
      endpoint => 'text-embedding-005'
    )).result AS embedding
  FROM UNNEST([
    'BigQuery is a serverless data warehouse',
    'Cloud Storage stores unstructured data as objects',
    'Vertex AI provides ML model training and serving',
    'Cloud Run deploys containerized applications',
    'Pub/Sub is a messaging service for event-driven systems'
  ]) AS doc
)
````

### Semantic search with `VECTOR_SEARCH`

`VECTOR_SEARCH` finds the closest matches from pre-computed embeddings. It requires embeddings stored in a table — create one with `AI.EMBED`, then search it.

**Hybrid search (Preview):** there is no separate hybrid search function. Hybrid — semantic similarity combined with lexical (keyword) matching — is a capability of the search functions above: pass `lexical_search_columns` and `lexical_search_query_value` to the single-query form of `VECTOR_SEARCH`, or set `mode => 'HYBRID'` on `AI.SEARCH`. The `distance` a hybrid search returns is a fused rank score rather than a cosine distance, so it is not comparable to semantic-only distances.

**How the fusion behaves.** Hybrid ranking is reciprocal rank fusion over 1-based ranks: `distance = 1 - ( 1/(60 + rank_vector) + 1/(61 + rank_lexical) )`, lowest first. The 60 is the canonical RRF constant over 1-based ranks; the lexical leg behaves as though its ranks start at 2, which is the same arithmetic as writing its base as 61. The lexical leg does not see the whole table: BigQuery hands it a candidate pool of the top `10 * top_k` rows by semantic rank, and a row deeper than that gets no lexical rank at all, however perfectly the token matches. Within that boundary the lexical leg ranks the entire candidate pool: BM25 matches take lexical ranks 1..m and every remaining pooled row falls back to its semantic order behind them, so **no pooled row forfeits a term**. Rank 1 in both legs is the best attainable score, `1 - (1/61 + 1/62) = 0.967478`. When nothing in the pool matches the lexical value, every row falls back to its semantic order in both legs, so each row at semantic rank `r` scores `1 - ( 1/(60 + r) + 1/(61 + r) )` — that is the zero-match signature, not a general rule for unmatched rows. Once any match fires, the matches take lexical ranks 1..m and each remaining row is pushed down behind them, so an unmatched row at semantic rank `r` carries `rank_lexical > r`. A lexical match promotes a row to lexical rank 1, worth `1/62` ≈ `0.0161` — enough to pull rows in from outside the semantic top-`top_k`, but only so deep. Two gates set how deep: the pool (`rank_vector <= 10 * top_k`) and the fused score (the promoted row must still beat the row holding the last slot). Effective reach is the smaller of the two — the score gate binds below `top_k => 51`, the pool gate at and above it — so a matched row can come from semantic rank 23 at `top_k => 10`, rank 110 at `top_k => 30`, rank 510 at `top_k => 51`, rank 640 at `top_k => 64`, and rank 1,000 at `top_k => 100`. So hybrid re-ranks *and* widens, but the widening is proportional to `top_k`, never unbounded: at small `top_k` an exact token only nudges rows already near the top, so do not expect recall from it; to reach a row sitting at semantic rank `R`, size `top_k` to at least `R/10`, and to more than that for targets in the first few hundred ranks, where the score gate binds instead. When the identifier is the whole query and you want certainty, use a predicate — `WHERE sku = @sku` cannot be outranked by a fusion score, and it has no pool. See [`functions/vector_search/`](functions/vector_search/), [`functions/ai_search/`](functions/ai_search/), and the [Catalog Search](workflows/catalog_search/) workflow.

````python
# Step 1: Create a table with pre-computed embeddings
embed_sql = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.overview_docs` AS
SELECT
  doc,
  (AI.EMBED(content => doc, endpoint => 'text-embedding-005')).result AS embedding
FROM UNNEST([
  'BigQuery is a serverless data warehouse',
  'Cloud Storage stores unstructured data as objects',
  'Vertex AI provides ML model training and serving',
  'Cloud Run deploys containerized applications',
  'Pub/Sub is a messaging service for event-driven systems'
]) AS doc
"""
client.query(embed_sql).result()
print('Embeddings table ready')

# Step 2: Search for the most relevant documents
search_sql = f"""
SELECT base.doc, distance
FROM VECTOR_SEARCH(
  TABLE `{PROJECT_ID}.{DATASET_ID}.overview_docs`,
  'embedding',
  (SELECT (AI.EMBED(
    content => 'How do I analyze large datasets?',
    endpoint => 'text-embedding-005'
  )).result AS embedding),
  top_k => 3,
  distance_type => 'COSINE'
)
ORDER BY distance
"""
client.query(search_sql).to_dataframe()
````

## 5. Document Processing — `AI.CHUNK_DOCUMENT`

`AI.CHUNK_DOCUMENT` (GA) is a **table-valued function** that splits unstructured documents in Cloud Storage (`PDF`, `HTML`, `DOCX`, `PPTX`, `XLSX`, `XLSM`) into layout-aware text chunks with ancestor section headings prepended — auto-provisioning Document AI's `bq-created-layout-parser` processor on first run with **zero `CREATE MODEL` DDL and zero object table required**.

- **Three input forms:** Pass a direct `'gs://bucket/path/*.pdf'` URI or wildcard literal (13 output columns), `TABLE project.dataset.object_table` (13 output columns), or a `(SELECT ..., OBJ.MAKE_REF(uri) AS ref FROM ...)` subquery (`5 + N` columns).
- **Target chunk size:** `chunk_size` accepts an `INT64` literal in `[1, 2000]` tokens (default `1000`).
- **Three-way `status` contract:** `status = ''` on valid chunks, a non-empty error string on unsupported formats (`.txt`, `.md`, standalone images) or size limits, and `NULL` when an `OBJ.MAKE_REF` URI points to a missing GCS object. Always filter downstream RAG pipelines with **`WHERE status = ''`**, which excludes both error strings and `NULL` rows.
- **When to use [`ML.PROCESS_DOCUMENT`](functions/ml_process_document/) instead:** Reach for `ML.PROCESS_DOCUMENT` when you need specialized Document AI entity extractors (Invoice, Expense, Form, OCR, Custom) or standalone raster images (`.png`, `.jpg`, `.tiff`).

**Deep dives:** [`functions/ai_chunk_document/`](functions/ai_chunk_document/) | [`functions/ml_process_document/`](functions/ml_process_document/) | [Document RAG workflow](workflows/document_rag/)

## 6. Predictive AI — `AI.FORECAST` and `AI.PREDICT`

Two zero-training foundation models cover prediction in BigQuery: **TimesFM** for time series and **TabFM** for tabular rows. Neither needs `CREATE MODEL`, a connection, or an endpoint.

`AI.FORECAST` uses Google's TimesFM foundation model to forecast time series data. No model training or creation needed — it works zero-shot on any time series.

**Pin the model version.** The TimesFM default changed from 2.0 to 2.5 with no release note, so an unpinned forecast silently returns different numbers than it used to. The call below passes `model => 'TimesFM 2.5'` explicitly; pinning is how a forecast stays stable across a future default move. `AI.FORECAST` is also reproducible run to run once pinned — but its two siblings are not. `AI.EVALUATE` and `AI.DETECT_ANOMALIES` return a different answer on a minority of runs even with `model` and `context_window` pinned and the query cache off (2 of 16 in the measured case, a ~32% swing in the reported error), so materialize those results once rather than re-running them.

**Deep dive:** [`functions/ai_forecast/`](functions/ai_forecast/)

````python
%%bigquery df_forecast --project {PROJECT_ID}

-- Generate a simple time series and forecast 7 days ahead
WITH daily_sales AS (
  SELECT
    DATE_ADD('2025-01-01', INTERVAL day_offset DAY) AS sale_date,
    -- Simulate sales with trend + weekly pattern + noise.
    -- The noise is a deterministic hash, not RAND(), so the plotting cell
    -- below can rebuild the exact same series that was forecast here.
    100 + (day_offset * 0.5)  -- upward trend
    + 20 * SIN(day_offset * 2 * ACOS(-1) / 7)  -- weekly cycle
    + (MOD(ABS(FARM_FINGERPRINT(CAST(day_offset AS STRING))), 1001) / 100 - 5)  -- noise in [-5, 5]
    AS daily_amount
  FROM UNNEST(GENERATE_ARRAY(0, 89)) AS day_offset
)
SELECT *
FROM AI.FORECAST(
  (SELECT * FROM daily_sales),
  data_col => 'daily_amount',
  timestamp_col => 'sale_date',
  horizon => 7,
  confidence_level => 0.95,
  model => 'TimesFM 2.5'  -- pin the version: the default flipped 2.0 -> 2.5
)
ORDER BY forecast_timestamp
````

````python
import matplotlib.pyplot as plt
import pandas as pd

# Rebuild the exact series that was forecast above (deterministic noise, same expression)
history_sql = f"""
SELECT
  DATE_ADD('2025-01-01', INTERVAL day_offset DAY) AS sale_date,
  100 + (day_offset * 0.5) + 20 * SIN(day_offset * 2 * ACOS(-1) / 7)
    + (MOD(ABS(FARM_FINGERPRINT(CAST(day_offset AS STRING))), 1001) / 100 - 5) AS daily_amount
FROM UNNEST(GENERATE_ARRAY(0, 89)) AS day_offset
ORDER BY sale_date
"""
df_history = client.query(history_sql).to_dataframe()

fig, ax = plt.subplots(figsize=(14, 5))

# History
ax.plot(df_history['sale_date'], df_history['daily_amount'],
        color='#4285F4', linewidth=1.5, label='Historical')

# Forecast
ax.plot(df_forecast['forecast_timestamp'], df_forecast['forecast_value'],
        color='#DB4437', linewidth=2, linestyle='--', label='Forecast')

# Confidence interval
ax.fill_between(
    df_forecast['forecast_timestamp'],
    df_forecast['prediction_interval_lower_bound'],
    df_forecast['prediction_interval_upper_bound'],
    color='#DB4437', alpha=0.1, label='95% Confidence Interval'
)

ax.set_title('AI.FORECAST — Zero-Shot Time Series Forecasting with TimesFM 2.5')
ax.set_xlabel('Date')
ax.set_ylabel('Daily Sales')
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

### Tabular prediction with `AI.PREDICT`

`AI.PREDICT` (Preview) runs regression and classification on tabular rows with the built-in TabFM foundation model. Pass it training rows and rows to predict — the **type of the label column picks the task**: a numeric label gives regression, a `BOOL` or `STRING` label gives classification. Features are limited to `STRING`, `BOOL`, `INT64`, `FLOAT64`, `NUMERIC`, and `BIGNUMERIC`, up to 50 feature columns.

**Deep dive:** [`functions/ai_predict/`](functions/ai_predict/)

````python
%%bigquery --project {PROJECT_ID}

-- Regression: predict penguin body mass from the other measurements.
-- A deterministic hash splits the 342 rows into ~302 training rows and a held-out bucket.
WITH penguins AS (
  SELECT *, MOD(ABS(FARM_FINGERPRINT(TO_JSON_STRING(t))), 10) AS bucket
  FROM `bigquery-public-data.ml_datasets.penguins` AS t
  WHERE body_mass_g IS NOT NULL
)
SELECT species, island, sex, body_mass_g AS actual_body_mass_g, predicted_body_mass_g
FROM AI.PREDICT(
  -- training rows: must contain the label column
  (SELECT * EXCEPT(bucket) FROM penguins WHERE bucket < 9),
  -- rows to predict: keeping the label here lets us compare actual vs predicted
  (SELECT * EXCEPT(bucket) FROM penguins WHERE bucket = 9 LIMIT 5),
  label_col => 'body_mass_g'
)
ORDER BY species
````

## 7. Augmented Analytics — `AI.KEY_DRIVERS`

`AI.KEY_DRIVERS` performs key driver / contribution analysis: it finds the data segments that drive a change in a summable metric between an interest set and a reference set. No model, connection, or endpoint — just a table with a metric, dimensions, and a BOOL interest/reference column.

**Deep dive:** [`functions/ai_key_drivers/`](functions/ai_key_drivers/)

````python
%%bigquery --project {PROJECT_ID}

-- Why did total trip duration change from April 2016 to April 2017?
SELECT
  ARRAY_TO_STRING(drivers, ', ') AS segment,
  difference,
  contribution
FROM AI.KEY_DRIVERS(
  (SELECT
     tripduration,
     usertype,
     gender,
     (EXTRACT(YEAR FROM starttime) = 2017) AS is_interest
   FROM `bigquery-public-data.new_york_citibike.citibike_trips`
   WHERE EXTRACT(MONTH FROM starttime) = 4
     AND EXTRACT(YEAR FROM starttime) IN (2016, 2017)
     AND tripduration IS NOT NULL),
  metric_col => 'tripduration',
  dimension_cols => ['usertype', 'gender'],
  interest_label_col => 'is_interest',
  top_k => 8
)
ORDER BY contribution DESC
````

## 9. SQL Ergonomics — GoogleSQL Reusable Column Aliases

Multi-step `AI.*` queries frequently chain scalar calls (`AI.TRANSLATE` -> `AI.CLASSIFY` -> `AI.IF`) or unpack `STRUCT`-returning functions (`AI.GENERATE`, `AI.EMBED`, `AI.TRANSLATE(..., return_details => TRUE)`) while filtering on `.status`. Historically, avoiding duplicate model calls per row required stacking `WITH` CTEs or lateral `UNNEST([...])` joins.

**GoogleSQL Reusable Column Aliases** (GA) let expressions in `SELECT`, `WHERE`, `GROUP BY`, and `HAVING` reference column aliases defined **earlier (to the left)** in the same `SELECT` list, evaluating each `AI.*` call once per row. Because changing identifier lookup can alter existing queries, BigQuery controls the feature via an opt-in regional project/organization option (`reusable_column_aliases_mode`, defaulting to `'off'` when unset).

First, inspect the effective setting in `region-us` (read-only — `0` rows means the option is unset and defaults to `'off'`):

````python
# Read-only inspection of the effective regional project option (never mutates project settings)
alias_option_sql = """
SELECT option_name, option_type, option_value
FROM `region-us`.INFORMATION_SCHEMA.EFFECTIVE_PROJECT_OPTIONS
WHERE option_name LIKE '%reusable_column_aliases%'
"""
df_alias_option = client.query(alias_option_sql).to_dataframe()
if df_alias_option.empty:
    print("0 rows returned: reusable_column_aliases_mode is unset in region-us (effective default: 'off').")
else:
    print(df_alias_option.to_string(index=False))
````

### Opt-in syntax, 5-tier name-resolution precedence, and limitations

> **Why this notebook does not execute `ALTER PROJECT`:** Enabling `reusable_column_aliases_mode` changes name resolution for **every** query executed in that region across the project (and requires `bigquery.config.update` / `resourcemanager.projects.update`). Review the precedence rules below before enabling it in a shared project.

```sql
-- Administrator opt-in for a region (NOT executed by this notebook; reset with = NULL or = 'off'):
ALTER PROJECT `your-gcp-project-id`
SET OPTIONS (`region-us.reusable_column_aliases_mode` = 'on');

-- 1. Lateral left-to-right AI.* chaining in one SELECT + WHERE (each AI.* call runs once per row):
SELECT
  raw_ticket,
  AI.TRANSLATE(raw_ticket, 'en', model => 'nmt') AS ticket_en,
  AI.CLASSIFY(ticket_en, ['Billing', 'Technical', 'Shipping']) AS category,
  AI.IF(CONCAT(ticket_en, ' — Does the customer report an outage or failed payment?')) AS is_urgent
FROM `your-gcp-project-id.bq_ai_functions.inbound_tickets`
WHERE is_urgent;

-- 2. Single-SELECT STRUCT unpacking + status filtering without a wrapper CTE:
SELECT
  ticket_id,
  AI.GENERATE(
    CONCAT('Extract root cause and 1-sentence summary: ', ticket_text),
    endpoint => 'gemini-2.5-flash',
    output_schema => 'root_cause STRING, summary STRING'
  ) AS ai_gen,
  ai_gen.root_cause,
  ai_gen.summary
FROM `your-gcp-project-id.bq_ai_functions.support_tickets`
WHERE ai_gen.status = '';
```

**Five-tier identifier resolution precedence (`reusable_column_aliases_mode = 'on'`):**

| Precedence | Scope | Shadowing Impact |
|------------|-------|------------------|
| **1 (Highest)** | `FROM` clause table column names | **Shadows `SELECT` aliases:** `SELECT AI.TRANSLATE(text, 'en') AS text, AI.CLASSIFY(text, ...)` passes the **untranslated base column `text`** into `AI.CLASSIFY` with no warning. Always choose a new alias name (`text_en`). |
| **2** | Reusable column aliases defined to the left in `SELECT` | **Shadows tiers 3–5 when `'on'`:** Enabling the option in an existing project can change queries where an outer correlated column, UDF parameter, or `DECLARE` variable shares a `SELECT` alias name. |
| **3** | Correlated column names from outer query scopes | Shadowed by any matching left-hand `SELECT` alias in the inner query. |
| **4** | SQL UDF and TVF parameter names | Shadowed by any matching left-hand `SELECT` alias inside the function body. |
| **5 (Lowest)** | Procedural language (`DECLARE`) variables | Shadowed by all four scopes above. |

*(Also avoid naming a `SELECT` struct alias after a `FROM` table alias `t` — `t.status` resolves against table `t` before struct `t`.)*

**Five documented limitations:**
1. **Strict left-to-right order in `SELECT`:** Forward references (`SELECT b + 1 AS a, 2 AS b`) fail validation, and self-references (`SELECT x + 1 AS x`) resolve `x` against Tier 1 `FROM` columns rather than the alias being defined.
2. **No aggregate-of-aggregate nesting in one `SELECT`:** `SELECT COUNT(*) AS cnt, MAX(cnt)` is invalid, and aggregate/window aliases cannot appear in `WHERE` (use `HAVING` / `QUALIFY`). Scalar post-processing of an aggregate alias — such as `AI.AGG(note, 'Summarize') AS brief, AI.TRANSLATE(brief, 'es', model => 'nmt') AS brief_es` — **is** supported.
3. **Not visible in `FROM` or `JOIN ... ON`:** Those clauses execute before `SELECT`.
4. **Not supported in `PIVOT`, `UNPIVOT`, or `MATCH_RECOGNIZE`.**
5. **Not supported in GoogleSQL pipe syntax (`|>`) or GQL (`GRAPH_TABLE`):** Pipe syntax *already* supports sequential left-to-right column extension natively via `|> EXTEND` with **no project option required** — making `|> EXTEND` (or a standard `WITH` CTE) the portable choice for notebooks and shared SQL templates.

**Reference:** [Reusable column aliases for unpacking `AI.*` outputs](reference/general-purpose-functions.md#reusable-column-aliases-for-unpacking-ai-outputs)

````python
%%bigquery --project {PROJECT_ID}

-- Portable today-runnable equivalent using GoogleSQL pipe syntax (no project option needed):
-- translate -> classify -> flag urgent tickets left-to-right, evaluating each AI.* call once per row
FROM UNNEST([
  'El pago con tarjeta falla cada vez que intento renovar nuestra suscripción.',
  'Bonjour, pouvez-vous envoyer une copie de la facture de mars ?',
  '本番環境のAPIゲートウェイが503エラーを返して停止しています。'
]) AS raw_ticket
|> EXTEND AI.TRANSLATE(raw_ticket, 'en', model => 'nmt') AS ticket_en
|> EXTEND
     AI.CLASSIFY(ticket_en, ['Billing', 'Technical', 'Account']) AS category,
     AI.IF(CONCAT(ticket_en, ' — Does the customer report an outage or failed payment?')) AS is_urgent
|> SELECT raw_ticket, ticket_en, category, is_urgent
|> ORDER BY is_urgent DESC, category
````
