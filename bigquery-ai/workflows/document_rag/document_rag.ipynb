# Document RAG — BigQuery AI Functions

A complete document-based Retrieval-Augmented Generation (RAG) pipeline built entirely in BigQuery SQL:

1. **Parse & Chunk** real PDFs into layout-aware sections with `AI.CHUNK_DOCUMENT` (plus parse-once chunk-size tuning via `AI.PARSE_DOCUMENT`)
2. **Embed** document chunks with `AI.EMBED` (`gemini-embedding-2`)
3. **Search** for relevant chunks with `VECTOR_SEARCH`
4. **Answer** questions with `AI.GENERATE` (`gemini-3.8-flash`), grounded in retrieved chunks with page and chunk citations

**What this demonstrates:**
- Building an end-to-end PDF RAG pipeline in BigQuery SQL with zero manual Document AI processor setup
- Layout-aware chunking (`chunk_size => 250`) across 20 single-page invoices and a multi-page policy document, preserving `chunk_id`, `start_page`, and `end_page` provenance
- Enforcing `AI.CHUNK_DOCUMENT`'s three-way `status` contract (`WHERE status = ''`) before embedding
- Comparing all four `AI.CHUNK_DOCUMENT` input modes: Object Table (`TABLE`), inline `OBJ.MAKE_REF` subquery, GCS URI wildcard, and pre-parsed `AI.PARSE_DOCUMENT(..., options => JSON '{"schema": "raw-docai"}')` (`parsed_document` — *"Parse Once, Chunk Many Times"*)
- Unified task-agnostic 3,072-dimensional embeddings (`gemini-embedding-2`, no `task_type` needed) and page-cited grounded generation vs ungrounded direct generation

**Functions used:** [`AI.CHUNK_DOCUMENT`](../../functions/ai_chunk_document/) | [`AI.PARSE_DOCUMENT`](../../functions/ai_parse_document/) | [`AI.EMBED`](../../functions/ai_embed/) | [`VECTOR_SEARCH`](../../functions/vector_search/) | [`AI.GENERATE`](../../functions/ai_generate/)

**Prerequisites:** [Setup guide](../../setup/) | [Document processing reference](../../reference/document-processing.md) | [Function reference](../../RESOURCES.md)

## Step 1 — Chunk documents with AI.CHUNK_DOCUMENT

`AI.CHUNK_DOCUMENT` reads documents from the object table, parses layout structure (headings, paragraphs, tables) via BigQuery's managed `bq-created-layout-parser` processor, and splits long documents into chunks up to `chunk_size` tokens (`1`–`2000`, default `1000`).

### Where `AI.CHUNK_DOCUMENT` and `AI.PARSE_DOCUMENT` fit alone and together

Both [`AI.CHUNK_DOCUMENT`](../../functions/ai_chunk_document/) and [`AI.PARSE_DOCUMENT`](../../functions/ai_parse_document/) use the same auto-provisioned `bq-created-layout-parser` processor (`$0` idle fee), so you can choose between one-step chunking, element-level JSON parsing, or chaining them together:

| Pattern | SQL Surface | Output Shape | When to Use It in Document Pipelines |
|---|---|---|---|
| **`AI.CHUNK_DOCUMENT` alone** *(used below for the main RAG index)* | `FROM AI.CHUNK_DOCUMENT(TABLE obj_table, connection_id => ..., chunk_size => 250)` | $N$ rows per document (`chunk_id`, `start_page`, `end_page`, `content`, `status`) | You already know your target `chunk_size` and want RAG-ready text chunks with page spans in a single SQL step |
| **`AI.PARSE_DOCUMENT` alone** (`"layout-v1"`, default) | `SELECT AI.PARSE_DOCUMENT(ref, connection_id => ...)` | 1 `JSON` value per document (`$.blocks` with `type`, `page_number`, `bbox`, HTML `<table>`) | You want to extract HTML `<table>` elements, filter specific block types (`heading`, `table`, `list_item`), or inspect normalized `[min_y, min_x, max_y, max_x]` coordinates |
| **`AI.PARSE_DOCUMENT` → `AI.CHUNK_DOCUMENT` together** (`"raw-docai"`, *shown at the end of Step 1*) | `AI.PARSE_DOCUMENT(ref, ..., options => JSON '{"schema": "raw-docai"}') AS parsed_document` → `FROM AI.CHUNK_DOCUMENT(TABLE parsed, chunk_size => ...)` | Cached `raw-docai` layout tree re-sliced into $N$ chunk rows at any `chunk_size` | **"Parse Once, Chunk Many Times"** — tune `chunk_size` (e.g. `150` vs `250` vs `500` tokens) or build multiple chunk tables without paying to re-parse the PDFs in Document AI |

Two operational details matter in every production pipeline:
1. **Matching `connection_id`:** When reading an object table (or `OBJ.MAKE_REF` subquery), `connection_id` in `AI.CHUNK_DOCUMENT` must match the connection bound to those object references (when reading a pre-parsed `parsed_document` table from `AI.PARSE_DOCUMENT`, `connection_id` is omitted on `AI.CHUNK_DOCUMENT`).
2. **Three-way `status` filtering:** `AI.CHUNK_DOCUMENT` sets `status = ''` (empty string) on success, a non-empty error message on corrupt or unsupported files, and `NULL` when an `ObjectRef` points to a missing GCS object. Always filter `WHERE status = ''` before downstream embedding — filtering `WHERE status IS NULL` would discard every valid chunk.

Below we materialize the raw chunking output with `chunk_size => 250`, verify the per-row `status` summary, and keep clean chunks for Step 2.

```python
raw_chunks = client.query(f"""
WITH raw AS (
  SELECT
    uri,
    chunk_id,
    start_page,
    end_page,
    content,
    status
  FROM AI.CHUNK_DOCUMENT(
    TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_objects`,
    connection_id => '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}',
    chunk_size => 250
  )
)
SELECT
  COUNT(*) AS total_rows,
  COUNTIF(status = '') AS valid_chunks,
  COUNTIF(status != '') AS parse_errors,
  COUNTIF(status IS NULL) AS missing_refs
FROM raw
""").to_dataframe()
print(
    f"Status check — total_rows: {raw_chunks.iloc[0]['total_rows']}, "
    f"valid_chunks (status = ''): {raw_chunks.iloc[0]['valid_chunks']}, "
    f"parse_errors: {raw_chunks.iloc[0]['parse_errors']}, "
    f"missing_refs: {raw_chunks.iloc[0]['missing_refs']}"
)

# Persist verified chunks (status = '') so downstream steps do not re-parse PDFs
client.query(f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_chunks` AS
SELECT
  uri,
  chunk_id,
  start_page,
  end_page,
  content
FROM AI.CHUNK_DOCUMENT(
  TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_objects`,
  connection_id => '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}',
  chunk_size => 250
)
WHERE status = ''
""").result()

stats = client.query(f"""
  SELECT
    COUNT(*) AS total_chunks,
    COUNT(DISTINCT uri) AS documents,
    MAX(chunk_id) AS max_chunks_in_doc,
    MAX(end_page) AS max_page_seen
  FROM `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_chunks`
""").to_dataframe()
row = stats.iloc[0]
print(
    f"Persisted {row['total_chunks']} chunks across {row['documents']} documents "
    f"(up to {row['max_chunks_in_doc']} chunks and page {row['max_page_seen']} in a single PDF)"
)
```

```python
# Inspect multi-page policy chunks alongside single-page invoice chunks
client.query(f"""
  SELECT
    REGEXP_EXTRACT(uri, r'[^/]+$') AS document,
    chunk_id,
    start_page,
    end_page,
    LEFT(REPLACE(content, '\\n', ' '), 160) AS content_preview
  FROM `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_chunks`
  ORDER BY
    CASE WHEN ENDS_WITH(uri, 'ap_vendor_policy.pdf') THEN 0 ELSE 1 END,
    uri,
    chunk_id
  LIMIT 8
""").to_dataframe()
```

### Alternative input modes: Inline `ObjectRef`, URI wildcard, and pre-parsed `AI.PARSE_DOCUMENT`

You don't need to create an external object table for ad-hoc document chunking or chunk-size experimentation. Beyond an object table, `AI.CHUNK_DOCUMENT` accepts three other input shapes:
- **An inline `ObjectRef` subquery** (`SELECT uri, OBJ.MAKE_REF(uri, 'connection') AS ref FROM ...`), which returns the 5 chunk columns (`chunk_id`, `start_page`, `end_page`, `content`, `status`) **plus** whatever passthrough columns your subquery selects (if you omit `uri` from the subquery `SELECT`, read `ref.uri` instead).
- **A Cloud Storage URI or wildcard string** (`'gs://bucket/path/*.pdf'`), which returns the same 13 columns as an object table.
- **A pre-parsed `parsed_document` `JSON` column** from [`AI.PARSE_DOCUMENT(..., options => JSON '{"schema": "raw-docai"}')`](../../functions/ai_parse_document/), which lets you parse PDFs once and re-chunk at multiple `chunk_size` values without re-invoking Document AI.

```python
# Ad-hoc inline ObjectRef subquery over 3 specific files — no object table required
sample_uris = [
    f'gs://{BUCKET}/{prefix}/ap_vendor_policy.pdf',
    f'gs://{BUCKET}/{prefix}/invoice_001.pdf',
    f'gs://{BUCKET}/{prefix}/invoice_003.pdf',
]
uris_sql = ', '.join(f"'{u}'" for u in sample_uris)

query = f"""
SELECT
  REGEXP_EXTRACT(uri, r'[^/]+$') AS document,
  chunk_id,
  start_page,
  end_page,
  LEFT(REPLACE(content, '\\n', ' '), 140) AS content_preview
FROM AI.CHUNK_DOCUMENT(
  (SELECT
    uri,
    OBJ.MAKE_REF(uri, '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}') AS ref
  FROM UNNEST([{uris_sql}]) AS uri),
  connection_id => '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}',
  chunk_size => 250
)
WHERE status = ''
ORDER BY document, chunk_id
LIMIT 6
"""
df = client.query(query).to_dataframe()
print(f'Inline ObjectRef approach: returned {len(df)} preview rows across {df["document"].nunique()} documents')
df
```

### Tuning `chunk_size` without re-parsing: `AI.PARSE_DOCUMENT` (`raw-docai`) → `AI.CHUNK_DOCUMENT`

When tuning RAG retrieval quality across a document corpus, calling `AI.CHUNK_DOCUMENT` directly on raw PDFs at three candidate `chunk_size` values invokes the Document AI Layout Parser three times on every page. Instead, use the two-stage **"Parse Once, Chunk Many Times"** pattern:

1. Call scalar [`AI.PARSE_DOCUMENT`](../../functions/ai_parse_document/) once per document with `options => JSON '{"schema": "raw-docai"}'` and alias the output column **`AS parsed_document`**. (`AI.PARSE_DOCUMENT`'s default `"layout-v1"` schema is built for direct SQL block/table inspection; `"raw-docai"` preserves the full Document AI layout tree required by `AI.CHUNK_DOCUMENT` — passing `"layout-v1"` into `AI.CHUNK_DOCUMENT` returns `status = 'INVALID_ARGUMENT: Unsupported document schema: layout-v1'`.)
2. Pass that table or CTE into `AI.CHUNK_DOCUMENT` at multiple `chunk_size` settings (`150`, `250`, and `500` tokens below) — `connection_id` is omitted on `AI.CHUNK_DOCUMENT` because Document AI has already parsed the PDF.

```python
# Parse the 3-page AP policy PDF ONCE with AI.PARSE_DOCUMENT (raw-docai),
# then compare AI.CHUNK_DOCUMENT across three chunk_size settings without re-parsing
policy_uri = f'gs://{BUCKET}/{prefix}/ap_vendor_policy.pdf'

query = f"""
WITH parsed_once AS (
  SELECT
    'ap_vendor_policy.pdf' AS document,
    AI.PARSE_DOCUMENT(
      OBJ.MAKE_REF('{policy_uri}', '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}'),
      connection_id => '{PROJECT_ID}.{LOCATION}.{CONNECTION_ID}',
      options => JSON '{{"schema": "raw-docai"}}'
    ) AS parsed_document
),
chunk_grid AS (
  SELECT 150 AS target_chunk_size, chunk_id, start_page, end_page, content, status
  FROM AI.CHUNK_DOCUMENT(TABLE parsed_once, chunk_size => 150)
  UNION ALL
  SELECT 250 AS target_chunk_size, chunk_id, start_page, end_page, content, status
  FROM AI.CHUNK_DOCUMENT(TABLE parsed_once, chunk_size => 250)
  UNION ALL
  SELECT 500 AS target_chunk_size, chunk_id, start_page, end_page, content, status
  FROM AI.CHUNK_DOCUMENT(TABLE parsed_once, chunk_size => 500)
)
SELECT
  target_chunk_size,
  COUNT(*) AS total_chunks,
  COUNTIF(status = '') AS valid_chunks,
  MIN(LENGTH(content)) AS min_chars,
  CAST(ROUND(AVG(LENGTH(content))) AS INT64) AS avg_chars,
  MAX(LENGTH(content)) AS max_chars,
  CONCAT(CAST(MIN(start_page) AS STRING), '-', CAST(MAX(end_page) AS STRING)) AS page_span
FROM chunk_grid
WHERE status = ''
GROUP BY target_chunk_size
ORDER BY target_chunk_size
"""
df_tuning = client.query(query).to_dataframe()
print('Parse-once chunk-size comparison on ap_vendor_policy.pdf (3 pages):')
df_tuning
```

## Step 2 — Embed chunks with AI.EMBED

Generate `3,072`-dimensional vectors for every persisted chunk using `gemini-embedding-2`. Because `gemini-embedding-2` produces unified, task-agnostic embeddings across queries and documents, `task_type` is omitted at both index and search time.

```python
query = f"""
CREATE OR REPLACE TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_embedded` AS
SELECT
  uri,
  chunk_id,
  start_page,
  end_page,
  content,
  (AI.EMBED(
    content => content,
    endpoint => 'gemini-embedding-2'
  )).result AS embedding
FROM `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_chunks`
WHERE content IS NOT NULL AND LENGTH(TRIM(content)) > 0
"""
client.query(query).result()

verify = client.query(f"""
  SELECT COUNT(*) AS chunks, ARRAY_LENGTH(embedding) AS dims
  FROM `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_embedded`
  GROUP BY dims
""").to_dataframe()
print(f'All {verify.iloc[0]["chunks"]} chunks embedded ({verify.iloc[0]["dims"]} dimensions)')
```

## Step 3 — Retrieve and generate (RAG)

Run the full RAG loop in a single SQL statement:
1. Embed the user question with `gemini-embedding-2`
2. Retrieve the top-`k` closest chunks with `VECTOR_SEARCH(..., distance_type => 'COSINE')`, carrying `uri`, `chunk_id`, `start_page`, and `end_page`
3. Format each chunk header with its document name, chunk number, and page span (`[doc | chunk N | pages S-E]`) and pass the assembled context to `AI.GENERATE` (`endpoint => 'gemini-3.8-flash'`) so the answer cites exact page provenance

```python
user_question = 'What is the total amount due on the invoice from Nexus Innovations Group?'

query = f"""
WITH retrieved AS (
  SELECT
    base.uri,
    base.chunk_id,
    base.start_page,
    base.end_page,
    base.content,
    distance
  FROM VECTOR_SEARCH(
    TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_embedded`,
    'embedding',
    query_value => (AI.EMBED(
      content => '{user_question}',
      endpoint => 'gemini-embedding-2'
    )).result,
    top_k => 3,
    distance_type => 'COSINE'
  )
),
context AS (
  SELECT STRING_AGG(
    CONCAT(
      '[', REGEXP_EXTRACT(uri, r'[^/]+$'),
      ' | chunk ', CAST(chunk_id AS STRING),
      ' | pages ', CAST(start_page AS STRING), '-', CAST(end_page AS STRING), '] ',
      content
    ),
    '\\n\\n'
    ORDER BY distance
  ) AS docs
  FROM retrieved
)
SELECT (AI.GENERATE(
  CONCAT(
    'You are a financial document analyst. Answer the user question using ONLY the provided document chunks. ',
    'Include exact invoice numbers, amounts, due dates, and cite each source as [filename, chunk N, pages S-E]. ',
    'If multiple matching invoices exist, report each one clearly.\\n\\n',
    'Document chunks:\\n', c.docs,
    '\\n\\nUser question: {user_question}'
  ),
  endpoint => 'gemini-3.8-flash'
)).result AS answer
FROM context c
"""
df = client.query(query).to_dataframe()
print(f'Question: {user_question}\n')
print(df.iloc[0]['answer'])
```

### View the retrieved chunks

Inspect the exact chunks and cosine distances `VECTOR_SEARCH` passed into the prompt above (lower cosine distance means higher semantic similarity).

```python
query = f"""
SELECT
  REGEXP_EXTRACT(base.uri, r'[^/]+$') AS document,
  base.chunk_id,
  base.start_page,
  base.end_page,
  ROUND(distance, 4) AS cosine_distance,
  LEFT(REPLACE(base.content, '\\n', ' '), 160) AS content_preview
FROM VECTOR_SEARCH(
  TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_embedded`,
  'embedding',
  query_value => (AI.EMBED(
    content => '{user_question}',
    endpoint => 'gemini-embedding-2'
  )).result,
  top_k => 3,
  distance_type => 'COSINE'
)
ORDER BY distance
"""
client.query(query).to_dataframe()
```

### Batch RAG: answer multiple questions across invoices and multi-page policy

`VECTOR_SEARCH` accepts a **query table** so BigQuery retrieves top-`k` chunks for an entire batch of questions in one pass. Notice how the first question pulls multi-page chunks (`chunk_id > 1`, `pages 2-3`) from `ap_vendor_policy.pdf`, while the invoice questions pull specific single-page invoices — and when a document genuinely omits a field, grounded prompting reports that cleanly instead of guessing.

```python
query = f"""
WITH questions AS (
  SELECT question_id, question
  FROM UNNEST([
    STRUCT(1 AS question_id, 'Under the Accounts Payable policy, what approval is required for invoices over $25,000 and how quickly must billing discrepancies be reported?' AS question),
    STRUCT(2 AS question_id, 'What are the payment terms on invoice INV-2024-0013, and what is the standard fallback term in the Accounts Payable policy?' AS question),
    STRUCT(3 AS question_id, 'List the invoices due in November 2023 from the retrieved chunks with their invoice numbers and total amounts.' AS question)
  ])
),
retrieved AS (
  SELECT
    query.question_id,
    query.question AS user_question,
    base.uri,
    base.chunk_id,
    base.start_page,
    base.end_page,
    base.content,
    distance
  FROM VECTOR_SEARCH(
    TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_embedded`,
    'embedding',
    (SELECT
       question_id,
       question,
       (AI.EMBED(
         content => question,
         endpoint => 'gemini-embedding-2'
       )).result AS embedding
     FROM questions),
    top_k => 4,
    distance_type => 'COSINE'
  )
),
context_per_question AS (
  SELECT
    question_id,
    user_question,
    STRING_AGG(
      CONCAT(
        '[', REGEXP_EXTRACT(uri, r'[^/]+$'),
        ' | chunk ', CAST(chunk_id AS STRING),
        ' | pages ', CAST(start_page AS STRING), '-', CAST(end_page AS STRING), '] ',
        content
      ),
      '\\n\\n'
      ORDER BY distance
    ) AS context
  FROM retrieved
  GROUP BY question_id, user_question
)
SELECT
  question_id,
  user_question,
  (AI.GENERATE(
    CONCAT(
      'Answer the question concisely based ONLY on the provided document chunks. ',
      'Cite every fact with [filename, chunk N, pages S-E]. ',
      'If a specific field is absent from an invoice, state that explicitly.\\n\\n',
      'Chunks:\\n', context,
      '\\n\\nQuestion: ', user_question
    ),
    endpoint => 'gemini-3.8-flash'
  )).result AS answer
FROM context_per_question
ORDER BY question_id
"""
df = client.query(query).to_dataframe()
for _, row in df.iterrows():
    print(f'Q{row["question_id"]}: {row["user_question"]}')
    print(f'A:  {row["answer"]}\n')
```

## Step 4 — RAG vs direct generation

Compare the exact same question answered **without** retrieved chunks vs **with** `AI.CHUNK_DOCUMENT` + `VECTOR_SEARCH` context on `gemini-3.8-flash`. Without retrieved evidence, the model cannot know private invoice totals or reference numbers; with RAG, every amount is anchored to a specific PDF chunk and page range.

```python
comparison_question = 'What is the total amount due on the invoice from Nexus Innovations Group, and whose sign-off does our AP policy require for the larger Nexus invoice?'

query = f"""
WITH no_rag AS (
  SELECT (AI.GENERATE(
    CONCAT(
      'Answer this question. If you do not have access to the private documents, say so clearly. ',
      'Question: {comparison_question}'
    ),
    endpoint => 'gemini-3.8-flash'
  )).result AS answer
),
rag_context AS (
  SELECT STRING_AGG(
    CONCAT(
      '[', REGEXP_EXTRACT(base.uri, r'[^/]+$'),
      ' | chunk ', CAST(base.chunk_id AS STRING),
      ' | pages ', CAST(base.start_page AS STRING), '-', CAST(base.end_page AS STRING), '] ',
      base.content
    ),
    '\\n\\n'
    ORDER BY distance
  ) AS docs
  FROM VECTOR_SEARCH(
    TABLE `{PROJECT_ID}.{DATASET_ID}.workflow_docrag_embedded`,
    'embedding',
    query_value => (AI.EMBED(
      content => '{comparison_question}',
      endpoint => 'gemini-embedding-2'
    )).result,
    top_k => 5,
    distance_type => 'COSINE'
  )
),
with_rag AS (
  SELECT (AI.GENERATE(
    CONCAT(
      'Answer this question based ONLY on the provided document chunks. ',
      'Include specific invoice numbers, dollar amounts, approval tier rules, and [filename, chunk N, pages S-E] citations.\\n\\n',
      'Document chunks:\\n', docs,
      '\\n\\nQuestion: {comparison_question}'
    ),
    endpoint => 'gemini-3.8-flash'
  )).result AS answer
  FROM rag_context
)
SELECT 1 AS sort_order, 'Without RAG (direct AI.GENERATE)' AS method, answer FROM no_rag
UNION ALL
SELECT 2 AS sort_order, 'With Document RAG (AI.CHUNK_DOCUMENT + VECTOR_SEARCH)' AS method, answer FROM with_rag
ORDER BY sort_order
"""
df = client.query(query).to_dataframe()
for _, row in df.iterrows():
    print(f'=== {row["method"]} ===')
    print(f'{row["answer"]}\n')
```

**Without RAG**, `gemini-3.8-flash` rightly declines because private vendor invoices and internal AP approval policies do not exist in pretraining weights. **With Document RAG**, a single SQL query combines facts across multiple PDFs — retrieving both Nexus Innovations Group invoices (`invoice_003.pdf` and `invoice_001.pdf`) and the relevant approval-threshold chunk from `ap_vendor_policy.pdf` — and cites the exact `chunk_id` and `start_page`–`end_page` for each claim.

### Choosing a document QA pattern in BigQuery

| Dimension | Direct `AI.GENERATE` (text only) | Multimodal `AI.GENERATE` (`ObjectRef` input) | Layout Parsing (`AI.PARSE_DOCUMENT`) | Document RAG (`AI.CHUNK_DOCUMENT` + `VECTOR_SEARCH`) |
|-----------|----------------------------------|----------------------------------------------|--------------------------------------|------------------------------------------------------|
| **Setup** | None | Object table or `OBJ.MAKE_REF` | Object table / `OBJ.MAKE_REF` (scalar `JSON` output) | Object table / `OBJ.MAKE_REF` (or `raw-docai` from `AI.PARSE_DOCUMENT`) + persisted chunk & embedding tables |
| **Per-question cost** | Lowest (prompt tokens only) | Re-reads full PDF bytes on every prompt | Parses once into `layout-v1` blocks or `raw-docai` JSON | Embeds query once + passes only top-`k` chunks |
| **Multi-document search** | Cannot access private files | Must already know which row/file to inspect | SQL filtering over `UNNEST(blocks)` (headings, HTML `<table>`, `bbox`) | Semantic/hybrid search across hundreds or thousands of PDFs, DOCX, PPTX, XLSX, and HTML files |
| **Citations** | None | File-level `uri` only | Exact `page_number`, `id`, and normalized `bbox` per block | Exact `uri`, `chunk_id`, `start_page`, and `end_page` |
| **Best for** | General world knowledge or text already inline in a SQL column | One-pass field extraction or visual inspection on a known document (see [Document Intelligence](../document_intelligence/)) | Extracting HTML tables, bounding boxes, or parsing once (`raw-docai`) to tune `AI.CHUNK_DOCUMENT` `chunk_size` | Reusable question answering and policy/invoice search across a growing document corpus |
